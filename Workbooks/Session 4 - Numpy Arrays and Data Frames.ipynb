## 1. Import the tools (3 minutes)

NumPy is commonly imported with the short name `np`. Pandas is commonly imported with the short name `pd`. These names let us call functions such as `np.array()` and `pd.DataFrame()` without writing the full library name each time.

```python
import numpy as np
import pandas as pd
```

## 2. Creating NumPy arrays (6 minutes)

A NumPy array stores values of the same general kind in an ordered structure. Arrays are useful when you want to perform the same numerical operation on many values. Create an array with `np.array()` and inspect its type and size.

```python
scores = np.array([72, 85, 91, 68, 88])

print(scores)
print(type(scores))
print('Number of scores:', scores.size)
print('Data type:', scores.dtype)
```

### Practice 1

Create an array called `temperatures` containing five whole-number temperatures. Print the array and its size.

## 3. Array positions and slices (6 minutes)

Like lists, NumPy arrays use zero-based indexing. A slice such as `scores[1:4]` selects positions 1, 2, and 3. The original array is not changed when you simply select values.

```python
print('First score:', scores[0])
print('Last score:', scores[-1])
print('Middle scores:', scores[1:4])

scores[0] = 75
print('Updated scores:', scores)
```

### Practice 2

Using `temperatures`, print the first value, the last two values, and then change the second value to `23`. Print the updated array.

```python
print('First temperature:', temperatures[0])
print('Last two temperatures:', temperatures[-2:])
temperatures[1] = 23
print('Updated temperatures:', temperatures)
```

## 4. Calculations on whole arrays (8 minutes)

Arithmetic with an array applies the operation to every element. This is called vectorized calculation. You do not need to write a `for` loop for simple operations such as adding 5 points or converting a score to a percentage.

```python
bonus_scores = scores + 5
percentages = scores / 100
doubled_scores = scores * 2

print('With bonus:', bonus_scores)
print('As proportions:', percentages)
print('Doubled:', doubled_scores)
```

NumPy also provides useful aggregate functions. `np.mean()` calculates an average, while `np.min()`, `np.max()`, and `np.sum()` find the smallest value, largest value, and total.

```python
print('Total:', np.sum(scores))
print('Average:', np.mean(scores))
print('Lowest:', np.min(scores))
print('Highest:', np.max(scores))
```

### Practice 3

Create `kilometers` from `[2, 5, 8, 10]`. Convert every distance to miles by multiplying by `0.621371`, then print the converted array and its average.

```python
kilometers = np.array([2, 5, 8, 10])
miles = kilometers * 0.621371
print('Miles:', miles)
print('Average miles:', np.mean(miles))
```

## 5. Conditions and filtering arrays (7 minutes)

A comparison with an array produces an array of Boolean values. This Boolean array can be used inside square brackets to select only the values that meet a condition.

```python
passed = scores >= 70
print('Passed mask:', passed)
print('Passing scores:', scores[passed])
print('Scores below 70:', scores[scores < 70])
print('Number passing:', np.sum(passed))
```

### Practice 4

Use `temperatures` to select values greater than `21`. Then calculate how many temperatures meet that condition.

```python
warm_days = temperatures > 21
print('Warm-day mask:', warm_days)
print('Warm temperatures:', temperatures[warm_days])
print('Number of warm days:', np.sum(warm_days))
```

## 6. Two-dimensional arrays (5 minutes)

An array can have rows and columns. The `shape` tells you the number of rows and columns. With two dimensions, use `table[row, column]` to select one value.

```python
weekly_scores = np.array([
    [72, 80, 75],
    [90, 88, 94],
    [65, 70, 68]
])

print(weekly_scores)
print('Shape:', weekly_scores.shape)
print('First students second score:', weekly_scores[0, 1])
print('First row:', weekly_scores[0])
print('Column averages:', np.mean(weekly_scores, axis=0))
```

## 7. Creating a pandas DataFrame (6 minutes)

A DataFrame is a table with labeled rows and columns. It is a good fit when each row represents one record and each column represents a variable. Create one from a dictionary where each key becomes a column name.

```python
students = pd.DataFrame({
    'name': ['Avery', 'Jordan', 'Morgan', 'Taylor', 'Riley'],
    'hours_studied': [4, 7, 3, 6, 5],
    'score': [72, 91, 68, 88, 84]
})

print(students)
```

Use `.head()` to preview rows and `.shape` to see the number of rows and columns. A column can be selected with square brackets, such as `students['score']`.

```python
print(students.head(3))
print('Shape:', students.shape)
print('Column names:', students.columns.tolist())
print('Scores:', students['score'].tolist())
print('Average score:', students['score'].mean())
```

### Practice 5

## 8. Filtering rows and adding columns (7 minutes)

A DataFrame can be filtered with a Boolean condition, just like a NumPy array. The result keeps only the rows where the condition is `True`. You can also create a new column by assigning a calculation to a new column name.

```python
high_scorers = students[students['score'] >= 85]
print('High scorers:')
print(high_scorers)

students['score_percent'] = students['score'] / 100
students['passed'] = students['score'] >= 70
print('Updated table:')
print(students)
```

### Practice 6

Filter `students` to find students who studied at least 5 hours. Add a column called `hours_per_point` by dividing `hours_studied` by `score`.

```python
regular_studiers = students[students['hours_studied'] >= 5]
students['hours_per_point'] = students['hours_studied'] / students['score']
print('Students with at least five study hours:')
print(regular_studiers)
print('Table with calculated column:')
print(students)
```

## 9. Mini-project: analyze class results (10 minutes)

Use the skills from this lesson to analyze the class results below. Your program should:

1. Create a NumPy array from the scores.
2. Print the class average, highest score, and number of passing students.
3. Create a DataFrame with each student's name, score, and attendance.
4. Add a `passed` column using a score of at least 70.
5. Print the students who passed and the students with attendance below 80.

Change some values after the first run and observe how the results change.

```python
names = ['Avery', 'Jordan', 'Morgan', 'Taylor', 'Riley', 'Casey']
class_scores = np.array([72, 91, 68, 88, 84, 59])
attendance = [95, 88, 76, 92, 81, 70]

print('Class average:', np.mean(class_scores))
print('Highest score:', np.max(class_scores))
print('Number passing:', np.sum(class_scores >= 70))

results = pd.DataFrame({
    'name': names,
    'score': class_scores,
    'attendance': attendance
})
results['passed'] = results['score'] >= 70

print('Passing students:')
print(results[results['passed']])
print('Students with attendance below 80:')
print(results[results['attendance'] < 80])
```

Extend the mini-project by adding a `grade` column. Use `A` for scores of 90 or more, `B` for scores from 80 to 89, `C` for scores from 70 to 79, and `Needs improvement` for lower scores. Then calculate the average attendance of students who passed.

```python
# Try the challenge here. One possible approach is to use np.select().
grade_conditions = [
    results['score'] >= 90,
    results['score'] >= 80,
    results['score'] >= 70
]
grade_choices = ['A', 'B', 'C']
results['grade'] = np.select(grade_conditions, grade_choices, default='Needs improvement')

print(results)
print('Average attendance for passing students:', results.loc[results['passed'], 'attendance'].mean())
```
