## 1. Your first Python expressions

Python can evaluate values and expressions directly. A string is text in quotes; numbers do not need quotes.

````python
"Hello, Python!"
2 + 3
10 * 4
````

## 2. Variables: giving values names

A variable is a name that refers to a value. Assignment uses one equals sign (`=`):

```python
name = value
```

Python figures out the value's type for you. Use clear, descriptive names with lowercase letters and underscores. A variable name cannot contain spaces or start with a number.

````python
student_name = "Alex"
age = 20
has_finished = False

print(student_name)
print(age)
print(has_finished)
````

## 3. Values and types

The most common beginner types are:

- `str`: text, such as `"hello"`
- `int`: whole numbers, such as `42`
- `float`: decimal numbers, such as `3.14`
- `bool`: `True` or `False`

Use `type()` to inspect a value.

````python
print(type(42))
print(type(3.14))
print(type("Python"))
print(type(True))
````

## 4. Arithmetic operators

| Operator | Meaning | Example |
| --- | --- | --- |
| `+` | add | `5 + 2` |
| `-` | subtract | `5 - 2` |
| `*` | multiply | `5 * 2` |
| `/` | divide, returns a float | `5 / 2` |
| `//` | floor division | `5 // 2` |
| `%` | remainder | `5 % 2` |
| `**` | exponent | `5 ** 2` |

````python
items = 17
people = 5

print(items + 3)
print(items - 3)
print(items * 2)
print(items / people)
print(items // people)
print(items % people)
print(2 ** 3)
````

## 5. Updating variables

A variable can be assigned a new value. The shorthand operators `+=`, `-=`, and `*=` update a variable using its current value.

````python
score = 10
score = score + 5
print(score)

score += 5
score *= 2
print(score)
````

## 6. Comparison operators

Comparisons produce a boolean value: `True` or `False`. Use `==` to ask whether two values are equal. A single `=` assigns a value.

`==`, `!=`, `>`, `<`, `>=`, and `<=` are the basic comparison operators.

````python
temperature = 22
print(temperature == 22)
print(temperature != 18)
print(temperature > 20)
print(temperature <= 25)
````

## 7. Boolean operators

Boolean operators combine or reverse conditions:

- `and`: both conditions must be true
- `or`: at least one condition must be true
- `not`: reverses `True` and `False`

````python
age = 19
has_ticket = True

can_enter = age >= 18 and has_ticket
needs_help = age < 18 or not has_ticket

print(can_enter)
print(needs_help)
````

## 8. Mini-project: plan a picnic

Use variables and operators to answer these questions:

1. How many sandwiches are needed if each person eats 2?
2. How many sandwiches are left if you have 10?
3. Is the picnic possible if it needs sunshine and the temperature is at least 18 degrees?

Change the values and run the cell again to test a different plan.

````python
people = 4
sandwiches_available = 10
sandwiches_per_person = 2
temperature = 21
is_sunny = True

sandwiches_needed = people * sandwiches_per_person
sandwiches_left = sandwiches_available - sandwiches_needed
picnic_is_possible = is_sunny and temperature >= 18

print("Sandwiches needed:", sandwiches_needed)
print("Sandwiches left:", sandwiches_left)
print("Picnic possible:", picnic_is_possible)
````

## 9. Common mistakes and recap

- Use quotes for text: `name = "Ada"`, not `name = Ada`.
- Use `==` for comparison and `=` for assignment.
- Check spelling: `temperature` and `temperture` are different names.
- Use parentheses when they make an expression easier to understand.
