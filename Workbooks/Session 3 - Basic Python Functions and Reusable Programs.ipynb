## 1. Why use functions? (5 minutes)

A function is a named group of instructions. Define it once with `def`, then call it whenever you need that behavior. The indented lines are the function body.

The code inside a function does not run when Python reads the definition. It runs when you call the function.

```python
def welcome_customer():
    print('Welcome to the classroom snack bar!')
    print('What would you like today?')

welcome_customer()
welcome_customer()
```

## 2. Parameters and arguments (7 minutes)

A **parameter** is a name in the function definition. An **argument** is the value you provide when calling the function. Parameters let one function work with different data.

## 3. Returning values (8 minutes)

A function can send a result back with `return`. Store that result in a variable or use it in another calculation. `print` displays a value; `return` gives a value back to the code that called the function.

```python
def calculate_total(order, menu):
    total = 0
    for item in order:
        total = total + menu[item]
    return total

menu = {'apple': 1, 'granola bar': 2, 'juice': 3}
order = ['apple', 'juice', 'apple']
subtotal = calculate_total(order, menu)
print('Subtotal:', subtotal, 'dollars')
print('After tax estimate:', subtotal * 1.05, 'dollars')
```

## 4. Default and keyword arguments (5 minutes)

A default argument is used when a caller leaves out a value. Keyword arguments make a call easier to read because they name the parameter being filled.

```python
def print_receipt(name, total, discount=0):
    final_total = total - discount
    print('Receipt for', name)
    print('Total:', final_total, 'dollars')

print_receipt('Sam', 6)
print_receipt(name='Riley', total=8, discount=1)
```

## 5. Scope: where names exist (4 minutes)

A variable created inside a function is local to that function. This keeps functions independent and prevents temporary calculation variables from accidentally changing the rest of your program. Pass information in through parameters and send results out with `return`.

```python
def make_discount(total):
    discount = 1 if total >= 5 else 0
    return discount

order_total = 7
discount_amount = make_discount(order_total)
print('Discount:', discount_amount)
# discount is local to make_discount; discount_amount is the returned result.
```

## 6. Validate input with a helper function (6 minutes)

Dictionaries have a `.get(key, fallback)` method. It returns the value for a key, or a fallback when the key is missing. This is useful for checking an order before looking up prices.

```python
def find_missing_items(order, menu):
    missing = []
    for item in order:
        if menu.get(item) is None:
            missing.append(item)
    return missing

menu = {'apple': 1, 'granola bar': 2, 'juice': 3}
order = ['apple', 'mystery snack', 'juice', 'mystery snack']
missing_items = find_missing_items(order, menu)
print('Missing:', missing_items)
```

## 7. List comprehensions (5 minutes)

A list comprehension is a compact way to build a list from a loop. Read `[expression for item in items]` as: create a new list containing the expression for every item. You can add an `if` condition at the end.

```python
menu = {'apple': 1, 'granola bar': 2, 'juice': 3, 'pretzels': 2}

inexpensive_snacks = [item for item in menu if menu[item] <= 2]
uppercase_snacks = [item.upper() for item in inexpensive_snacks]

print('Inexpensive snacks:', inexpensive_snacks)
print('Display labels:', uppercase_snacks)
```

## 8. Capstone: build a reusable checkout (10 minutes)

Use your functions to create a checkout workflow. The program should:

1. Check whether all requested items are available.
2. Calculate the subtotal.
3. Apply a 1 dollar discount for totals of at least 5 dollars.
4. Apply an additional coupon discount when `has_coupon` is `True`.
5. Print a receipt from one final function.

Try changing the order, name, and coupon value after you run the solution.

```python
def calculate_discount(subtotal, has_coupon=False):
    discount = 1 if subtotal >= 5 else 0
    if has_coupon:
        discount = discount + 1
    return discount

def checkout(name, order, menu, has_coupon=False):
    subtotal = calculate_total_if_available(order, menu)
    if subtotal is None:
        return

    discount = calculate_discount(subtotal, has_coupon)
    print_receipt(name, subtotal, discount=discount)
    print('Items:', order)

menu = {'apple': 1, 'granola bar': 2, 'pretzels': 2, 'juice': 3}
checkout('Sam', ['granola bar', 'juice', 'apple'], menu, has_coupon=True)
checkout('Taylor', ['apple', 'unknown item'], menu)
```
