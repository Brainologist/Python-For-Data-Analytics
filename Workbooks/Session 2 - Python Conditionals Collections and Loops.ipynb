## 1. Making comparisons

A comparison produces a Boolean value: either `True` or `False`. Common comparison operators are `==` (equal to), `!=` (not equal to), `>` and `<`, plus `>=` and `<=`.

```python
price = 3
money = 5

print(price < money)
print(price == 3)
print(money != 3)
```

## 2. Conditional logic

An `if` statement runs code only when a condition is true. Add `elif` for another possibility and `else` for everything left over. Indentation tells Python which lines belong to each branch.

```python
money = 5
snack_price = 3

if money >= snack_price:
    print('You can buy the snack.')
elif money > 0:
    print('You have some money, but not enough yet.')
else:
    print('You need to find some money first.')
```

### Combining conditions

Use `and` when both conditions must be true, `or` when either condition is enough, and `not` to reverse a Boolean value.

```python
is_lunchtime = True
has_money = True

if is_lunchtime and has_money:
    print('The snack bar is ready for your order.')

allergy_safe = False
vegetarian = True
if allergy_safe or vegetarian:
    print('Show the customer the ingredient information.')

if not allergy_safe:
    print('Ask about allergies before serving the item.')
```

## 3. Lists: keeping an ordered collection

A list stores multiple values in order. Positions start at index `0`, so the first item is `snacks[0]`. Lists can grow and change.

```python
snacks = ['apple', 'granola bar', 'pretzels']
print(snacks)
print('First snack:', snacks[0])
print('Last snack:', snacks[-1])

snacks.append('juice')
snacks.remove('pretzels')
print('Updated snacks:', snacks)
print('Number of snacks:', len(snacks))
```

## 4. Dictionaries: labeled information

A dictionary stores key/value pairs. Keys act like labels, which is useful when each item has several properties.

```python
prices = {
    'apple': 1,
    'granola bar': 2,
    'pretzels': 2,
    'juice': 3
}

print('A juice costs', prices['juice'], 'dollars.')
prices['juice'] = 2
prices['banana'] = 1
print('Updated prices:', prices)
```

## 5. `for` loops: repeat once per item

A `for` loop visits each item in a collection. The indented block runs once for each item.

```python
orders = ['apple', 'juice', 'apple', 'granola bar']

for item in orders:
    print('Preparing:', item)

print('Items in order:', len(orders))
```

### `range` and totals

`range(4)` gives the numbers `0, 1, 2, 3`. A loop can use those numbers to repeat an action a set number of times.

```python
total = 0
prices = {'apple': 1, 'juice': 2, 'granola bar': 2}
orders = ['apple', 'juice', 'apple']

for item in orders:
    total = total + prices[item]

print('Order total:', total, 'dollars')

for number in range(3):
    print('Receipt line', number + 1)
```

### Practice

Use a `for` loop to count how many times the word `apple` appears in `orders`. Start with `apple_count = 0`, then increase it inside the loop when the current item is an apple.

```python
orders = ['apple', 'juice', 'apple', 'granola bar']
apple_count = 0

for item in orders:
    if item == 'apple':
        apple_count = apple_count + 1

print('Apples ordered:', apple_count)
```

## 6. `while` loops: repeat while a condition is true

A `while` loop is useful when you do not know exactly how many repetitions you need. Make sure something changes inside the loop so the condition eventually becomes false.

```python
snacks_left = 3

while snacks_left > 0:
    print('Serving a snack. Snacks left:', snacks_left)
    snacks_left = snacks_left - 1

print('The tray is empty.')
```

## 7. Mini-project: build a snack bar checkout

Put the ideas together. Your program should:

1. Store snack prices in a dictionary.
2. Store a student's order in a list.
3. Use a loop to calculate the total.
4. Use conditional logic to apply a 1 dollar discount when the total is at least 5 dollars.
5. Print a friendly receipt.

Start by filling in or changing the values below. Then run the solution cell.

```python
menu = {'apple': 1, 'granola bar': 2, 'pretzels': 2, 'juice': 3}
order = ['granola bar', 'juice', 'apple']
student_name = 'Sam'

subtotal = 0
for item in order:
    subtotal = subtotal + menu[item]

if subtotal >= 5:
    discount = 1
else:
    discount = 0

total = subtotal - discount
print('Receipt for', student_name)
print('Items:', order)
print('Subtotal:', subtotal, 'dollars')
print('Discount:', discount, 'dollars')
print('Total:', total, 'dollars')
```
